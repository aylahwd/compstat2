## 1. Introduction to NumPy

### 1.1 What Is NumPy?

**NumPy** (*Numerical Python*) is a Python library designed for numerical and scientific computation. It makes it convenient to organize numerical data, perform operations on entire arrays, and prepare data for statistical analysis, scientific modeling, and machine learning.

### 1.2 Importing NumPy

The conventional alias for NumPy is **`np`**. If NumPy is not installed yet, remove the `#` in the first cell below and run it.

```python
# Install NumPy (only if it is not installed yet): remove the '#' below
# %pip install numpy
```

```python
import numpy as np

print("NumPy version:", np.__version__)
```

### 1.3 Python Lists vs. NumPy Arrays

A **Python list** can store elements of different types.

A **NumPy array** stores elements of a single, common type, which makes numerical computation more efficient.

The same operator can behave differently: `+` *concatenates* two lists, but *adds corresponding elements* of two NumPy arrays.

```python
python_list = [2, 4, 6]
numpy_array = np.array([2, 4, 6])

print("Python list:", python_list)
print("NumPy array:", numpy_array)
print("Python type:", type(python_list))
print("NumPy type: ", type(numpy_array))
```

```python
print(python_list + python_list)   # concatenation
print(numpy_array + numpy_array)   # element-wise addition
```

## 2. Creating NumPy Arrays

### 2.1 Creating Arrays from Python Lists

`np.array()` converts Python data into a NumPy array. The number of dimensions is determined by the structure of the data:

- **1D array:** a vector.
- **2D array:** a matrix (rows and columns).
- **3D array:** a stack of 2D matrices.
- **4D array:** a collection of 3D arrays.

The input must have a consistent **shape**, which is the number of elements along each dimension.

### 2.2 One-Dimensional Array (1D)

A 1D array represents measurements of a **single variable**, such as the scores of five students.

```python
student_scores = [75, 80, 92, 86, 88]
scores_1d = np.array(student_scores)

print(scores_1d)
print("Dimensions:", scores_1d.ndim)
print("Shape:", scores_1d.shape)
```

### 2.3 Two-Dimensional Array (2D)

A 2D array represents data **in rows and columns**. In the example below, each row is a student and each column is a different assessment.

```python
student_assessments = [[80, 85, 90],
                       [75, 88, 84],
                       [92, 90, 95]]

scores_2d = np.array(student_assessments)

print(scores_2d)
print("Dimensions:", scores_2d.ndim)
print("Shape:", scores_2d.shape)
```

The shape `(3, 3)` indicates **three rows and three columns**.

### 2.4 Three-Dimensional Array (3D)

A 3D array can be understood as **a stack of matrices**.

```python
week_1 = np.array([[10, 12],
                   [14, 16]])
week_2 = np.array([[11, 13],
                   [15, 17]])

observations_3d = np.array([week_1, week_2])

print(observations_3d)
print("Dimensions:", observations_3d.ndim)
print("Shape:", observations_3d.shape)
```

The shape `(2, 2, 2)` indicates **two matrices**, each with **two rows and two columns**. The first axis selects which matrix.

### 2.5 Four-Dimensional Array (4D)

A 4D array adds another level of organization: think of it as a collection of 3D arrays. Such structures appear in scientific data and batches of multidimensional observations.

```python
first_cube = np.array([[[1, 2], [3, 4]],
                       [[5, 6], [7, 8]]])
second_cube = np.array([[[11, 12], [13, 14]],
                        [[15, 16], [17, 18]]])

observations_4d = np.array([first_cube, second_cube])

print(observations_4d)
print("Dimensions:", observations_4d.ndim)
print("Shape:", observations_4d.shape)
```

### 2.6 Creating Arrays from Scratch: `np.zeros()`

`np.zeros()` creates an array filled with zeros. Pass an integer for a 1D array, or a tuple for a multidimensional array.

```python
zero_vector = np.zeros(5)
zero_matrix = np.zeros((3, 4))

print("1D zeros:")
print(zero_vector)
print("2D zeros:")
print(zero_matrix)
```

### 2.7 Random Arrays: `np.random.random()`

`np.random.random()` generates random decimal numbers from **0 up to (but not including) 1**. Setting a `seed` makes the random numbers identical every time the code is run.

```python
np.random.seed(42)
random_measurements = np.random.random((2, 3))

print(random_measurements)
print("Shape:", random_measurements.shape)
```

### 2.8 Sequences: `np.arange()`

`np.arange(start, stop, step)` creates a sequence of numbers.

- `start`: first value (default `0`).
- `stop`: upper limit, **not included**.
- `step`: spacing between values (default `1`).

```python
sequence_a = np.arange(5)
sequence_b = np.arange(2, 11, 2)
sequence_c = np.arange(-3, 4, 3)

print("Default sequence:     ", sequence_a)
print("Even-number sequence: ", sequence_b)
print("Stepped sequence:     ", sequence_c)
```

## 3. Array Attributes and Data Types

### 3.1 Dimensionality and Axes

An **axis** is a direction along a dimension of the array. In a 2D array, `axis=0` runs down the rows and `axis=1` runs across the columns. A 3D array has three axes: layer, row, and column.

| Attribute | Meaning | Example result |
|---|---|---|
| `.ndim` | Number of dimensions (axes) | `2` |
| `.shape` | Length along each axis | `(3, 4)` |
| `.size` | Total number of elements | `12` |
| `.dtype` | Common data type of the elements | `int64`, `float64`, etc. |

```python
measurement_matrix = np.array([[12, 15, 18, 21],
                               [14, 16, 20, 22],
                               [11, 13, 17, 19]])

print("Array:")
print(measurement_matrix)
print("ndim: ", measurement_matrix.ndim)
print("shape:", measurement_matrix.shape)
print("size: ", measurement_matrix.size)
print("dtype:", measurement_matrix.dtype)
```

### 3.2 Understanding Shape and Size

A shape of `(3, 4)` means three rows and four columns, so the size is `3 × 4 = 12` elements. For 3D and 4D arrays, the size is the product of all entries in the shape.

```python
print("1D:", scores_1d.ndim, scores_1d.shape, scores_1d.size)
print("2D:", scores_2d.ndim, scores_2d.shape, scores_2d.size)
print("3D:", observations_3d.ndim, observations_3d.shape, observations_3d.size)
print("4D:", observations_4d.ndim, observations_4d.shape, observations_4d.size)
```

### 3.3 Common NumPy Data Types (`dtype`)

NumPy supports integers (`int`), floating-point numbers (`float`), Booleans (`bool`), and strings (`str`). The `dtype` NumPy reports gives more detail, such as the number of bits used per value.

Because a NumPy array uses one common data type, mixing different kinds of values triggers automatic type conversion (see Section 3.6).

```python
integer_values = np.array([1, 2, 3])
float_values = np.array([1.5, 2.5, 3.5])
boolean_values = np.array([True, False, True])
string_values = np.array(["low", "medium", "high"])

print("Integer:", integer_values.dtype)
print("Float:  ", float_values.dtype)
print("Boolean:", boolean_values.dtype)
print("String: ", string_values.dtype)
```

### 3.4 Specifying a Data Type Explicitly

Use the `dtype` argument of `np.array()` to request a specific type. A smaller numeric type may use less memory, but it has a more limited range or precision.

```python
measurements_float32 = np.array([1.25, 2.75, 3.50], dtype=np.float32)
counts_int32 = np.array([10, 20, 30], dtype=np.int32)

print(measurements_float32, measurements_float32.dtype)
print(counts_int32, counts_int32.dtype)
```

**Caution:** requesting an integer `dtype` for decimal values **truncates** the fractional part (it does not round).

```python
measurements_int32 = np.array([1.25, 2.75, 3.50], dtype=np.int32)
measurements_int32
```

### 3.5 Type Conversion with `astype()`

`.astype()` returns a copy of the array converted to the specified type. A common use is converting strings that contain valid numbers into numeric values.

Be careful: converting floats to integers discards the fractional part; it is not the same as rounding to the nearest integer.

```python
text_scores = np.array(["70", "85", "90"])
numeric_scores = text_scores.astype(np.int64)

print("Before:", text_scores, text_scores.dtype)
print("After: ", numeric_scores, numeric_scores.dtype)
```

```python
continuous_values = np.array([1.9, 2.1, 3.8])
integer_conversion = continuous_values.astype(int)

print("Original: ", continuous_values)
print("Converted:", integer_conversion)
```

### 3.6 Type Coercion

**Type coercion** happens when NumPy automatically chooses a common data type for all elements of an array. For example, integers become floats when a decimal is included, and numbers become strings when mixed with text.

> **Always inspect `.dtype` before numerical analysis.**

```python
coercion_a = np.array([10, 20.5, 30])
coercion_b = np.array([True, False, 5])
coercion_c = np.array([1, "two", 3])

print(coercion_a, coercion_a.dtype)
print(coercion_b, coercion_b.dtype)
print(coercion_c, coercion_c.dtype)
```

## 4. Reshaping Arrays

### 4.1 Understanding Array Shape

The **shape** describes how elements are organized. A 1D array with six elements has shape `(6,)`, not `(1, 6)`. A row-shaped 2D array and a column-shaped 2D array have different shapes even when they hold the same six values.

```python
base_values = np.arange(1, 7)

row_array = base_values.reshape(1, 6)
column_array = base_values.reshape(6, 1)

print("1D shape:", base_values.shape)
print("Row shape:", row_array.shape)
print(row_array)
print("Column shape:", column_array.shape)
print(column_array)
```

### 4.2 Reshaping with `reshape()`

`.reshape()` rearranges the elements **without changing how many there are**. A 1D array of 12 values can become `(3, 4)`, `(4, 3)`, or `(2, 2, 3)`, because each shape has exactly 12 positions.

NumPy reads the elements in their existing order and fills the new shape.

```python
values_12 = np.arange(1, 13)

matrix_3x4 = values_12.reshape(3, 4)
matrix_4x3 = values_12.reshape(4, 3)
array_2x2x3 = values_12.reshape(2, 2, 3)

print("Original:", values_12)
print("3 x 4:")
print(matrix_3x4)
print("4 x 3:")
print(matrix_4x3)
print("2 x 2 x 3:")
print(array_2x2x3)
```

### 4.3 Reshape Compatibility

The total number of elements **must match** the requested shape. For example, an array of size 12 cannot become `(5, 3)`, because that shape needs 15 elements.

NumPy can calculate **one** missing dimension if you pass `-1`.

```python
reshaped_automatically = values_12.reshape(3, -1)

print(reshaped_automatically)
print("Inferred shape:", reshaped_automatically.shape)
```

### 4.4 Flattening with `flatten()`

`.flatten()` converts a multidimensional array into a **new 1D copy**. It is useful when an analysis needs all elements in a single sequence.

```python
original_matrix = np.array([[1, 2],
                            [5, 7],
                            [6, 6]])
flattened_values = original_matrix.flatten()

print("Original:")
print(original_matrix)
print("Flattened:")
print(flattened_values)
print("Flattened shape:", flattened_values.shape)
```

### 4.5 `reshape()` vs. `flatten()`

| Operation | Main purpose | Dimensions after operation |
|---|---|---|
| `reshape(new_shape)` | Rearrange elements into a chosen shape | Determined by `new_shape` |
| `flatten()` | Convert all elements into a new 1D array | Always 1D |

## 5. Basic Array Indexing and Slicing

### 5.1 Indexing a 1D Array

**Indexing** selects an element by its position. Python uses **zero-based indexing**: index `0` is the first element, `1` is the second, and so on.

Negative indices count backward from the end: `-1` is the last element.

```python
values = np.array([2, 4, 6, 8, 10])

print("First element: ", values[0])
print("Fourth element:", values[3])
print("Last element:  ", values[-1])
```

### 5.2 Indexing a 2D Array

A 2D array needs a **row index** and a **column index**. The syntax is `array[row, column]`, and both indices start from zero.

```python
marks = np.array([[80, 85, 90],
                  [75, 88, 84],
                  [92, 90, 95]])

print(marks)
print("Row 1, column 2:", marks[1, 2])
print("Row 2, column 0:", marks[2, 0])
```

### 5.3 Selecting Rows and Columns

A single index selects a whole row (`array[row]`). Combine `:` with an index to select a whole column (`array[:, column]`). The colon means **all positions** along that axis.

```python
print("First row:   ", marks[0])
print("Second row:  ", marks[1, :])
print("Third column:", marks[:, 2])
print("First column:", marks[:, 0])
```

### 5.4 Slicing a 1D Array

**Slicing** retrieves several consecutive elements. The syntax is `array[start:stop]`.

- `start` is included.
- `stop` is excluded.
- An omitted `start` means the beginning of the array.
- An omitted `stop` means the end of the array.

```python
sequence = np.array([10, 20, 30, 40, 50, 60])

print("Positions 1 to 3:      ", sequence[1:4])
print("First three:           ", sequence[:3])
print("From position 3 onward:", sequence[3:])
print("All elements:          ", sequence[:])
```

### 5.5 Slicing a 2D Array

A 2D slice specifies a range of rows and a range of columns: `array[row_start:row_stop, col_start:col_stop]`. This is how you extract a submatrix.

```python
number_grid = np.arange(1, 26).reshape(5, 5)

print("Original matrix:")
print(number_grid)
print("Rows 1-3, columns 2-4:")
print(number_grid[1:4, 2:5])
```

### 5.6 Slicing with Steps

A slice can take a third argument: `array[start:stop:step]`. The `step` is the interval between selected positions.

- `::2` selects every second element, starting at the beginning.
- `1::2` selects every second element, starting at index 1.
- `[::-1]` reverses a 1D array.

Steps can also be applied separately to the row and column axes of a 2D array.

```python
print("Every second element:      ", sequence[::2])
print("Every second from index 1: ", sequence[1::2])
print("Reversed:                  ", sequence[::-1])
print("Every second row and column:")
print(number_grid[::2, ::2])
```
